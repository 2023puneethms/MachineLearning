```python
import pandas as pd

# ============================================================
# FIND-S ALGORITHM
# ============================================================

# ------------------------------------------------------------
# 1. Read the dataset
# ------------------------------------------------------------

file_path = "play_tennis.csv"

data = pd.read_csv(file_path)

# Remove unnecessary spaces from column names
data.columns = data.columns.str.strip()

print("================================================")
print("DATASET")
print("================================================")
print(data)

# ------------------------------------------------------------
# 2. Display column names
# ------------------------------------------------------------

print("\n================================================")
print("COLUMN NAMES")
print("================================================")

print(data.columns.tolist())

# ------------------------------------------------------------
# 3. Automatically identify target column
# ------------------------------------------------------------
# We assume the LAST column is the target/class column.

target_column = data.columns[-1]

print("\nTarget Column:", target_column)

# All columns except target are attributes
attributes = list(data.columns[:-1])

print("\nAttributes:")
print(attributes)

# ------------------------------------------------------------
# 4. Display unique values of target column
# ------------------------------------------------------------

print("\nTarget Values:")
print(data[target_column].unique())

# ------------------------------------------------------------
# 5. Initialize the most specific hypothesis
# ------------------------------------------------------------

hypothesis = ["Ø"] * len(attributes)

print("\n================================================")
print("INITIAL HYPOTHESIS")
print("================================================")

print("<" + ", ".join(hypothesis) + ">")

# ------------------------------------------------------------
# 6. Function to identify positive examples
# ------------------------------------------------------------

def is_positive(value):

    value = str(value).strip().lower()

    positive_values = [
        "yes",
        "y",
        "true",
        "1",
        "positive",
        "play",
        "played"
    ]

    return value in positive_values


# ------------------------------------------------------------
# 7. Apply Find-S Algorithm
# ------------------------------------------------------------

print("\n================================================")
print("FIND-S PROCESS")
print("================================================")

for index, row in data.iterrows():

    target_value = row[target_column]

    # --------------------------------------------------------
    # Process only positive examples
    # --------------------------------------------------------

    if is_positive(target_value):

        print("\nExample", index + 1, "- POSITIVE")

        print("Attributes:",
              row[attributes].tolist())

        # Compare each attribute with hypothesis
        for i, attribute in enumerate(attributes):

            current_value = row[attribute]

            # First positive example
            if hypothesis[i] == "Ø":

                hypothesis[i] = current_value

            # Different values -> generalize to ?
            elif hypothesis[i] != current_value:

                hypothesis[i] = "?"

        print("Current Hypothesis:")
        print("<" + ", ".join(map(str, hypothesis)) + ">")

    else:

        print("\nExample", index + 1, "- NEGATIVE")
        print("Attributes:",
              row[attributes].tolist())

        print("Negative example is ignored.")


# ------------------------------------------------------------
# 8. Display final hypothesis
# ------------------------------------------------------------

print("\n================================================")
print("FINAL MOST SPECIFIC HYPOTHESIS")
print("================================================")

print("<" + ", ".join(map(str, hypothesis)) + ">")

print("\nFind-S algorithm completed successfully.")
```
