```python
# ============================================================
# DATA LOADING, CLEANING AND EXPLORATORY DATA ANALYSIS (EDA)
# Dataset: Titanic Dataset
# Libraries: NumPy, Pandas, Matplotlib
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. LOAD THE DATASET
# ------------------------------------------------------------

# Make sure train.csv is in the same folder as this Python file
df = pd.read_csv("train.csv")

print("Dataset loaded successfully!\n")

# Display first 5 rows
print("First 5 rows:")
print(df.head())

# Display last 5 rows
print("\nLast 5 rows:")
print(df.tail())


# ------------------------------------------------------------
# 2. BASIC DATA INFORMATION
# ------------------------------------------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()


# ------------------------------------------------------------
# 3. CHECK FOR MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values in Each Column:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)


# ------------------------------------------------------------
# 4. HANDLE MISSING VALUES
# ------------------------------------------------------------

# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin contains many missing values.
# Instead of deleting many rows, replace missing values
# with "Unknown".
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("\nMissing values after cleaning:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 5. REMOVE DUPLICATE RECORDS
# ------------------------------------------------------------

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDataset shape after removing duplicates:")
print(df.shape)


# ------------------------------------------------------------
# 6. NUMPY OPERATIONS
# ------------------------------------------------------------

# Convert selected numerical columns into NumPy array
numeric_data = df[["Age", "Fare", "SibSp", "Parch"]].to_numpy()

print("\nNumPy Array:")
print(numeric_data[:5])

# Mean using NumPy
print("\nMean Age using NumPy:")
print(np.mean(df["Age"]))

# Maximum Fare
print("\nMaximum Fare using NumPy:")
print(np.max(df["Fare"]))

# Minimum Fare
print("\nMinimum Fare using NumPy:")
print(np.min(df["Fare"]))

# Standard deviation of Age
print("\nStandard Deviation of Age:")
print(np.std(df["Age"]))


# ------------------------------------------------------------
# 7. SUMMARY STATISTICS
# ------------------------------------------------------------

print("\nSummary Statistics:")
print(df.describe())

print("\nSummary Statistics including categorical columns:")
print(df.describe(include="all"))


# ------------------------------------------------------------
# 8. UNIQUE VALUES / CATEGORICAL ANALYSIS
# ------------------------------------------------------------

print("\nUnique values in Sex:")
print(df["Sex"].unique())

print("\nNumber of passengers by Sex:")
print(df["Sex"].value_counts())

print("\nNumber of passengers in each class:")
print(df["Pclass"].value_counts())

print("\nNumber of survivors and non-survivors:")
print(df["Survived"].value_counts())


# ------------------------------------------------------------
# 9. DISTRIBUTION OF AGE
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20, edgecolor="black")

plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.grid(axis="y", alpha=0.3)
plt.show()


# ------------------------------------------------------------
# 10. DISTRIBUTION OF FARE
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(df["Fare"], bins=30, edgecolor="black")

plt.title("Distribution of Passenger Fare")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.grid(axis="y", alpha=0.3)
plt.show()


# ------------------------------------------------------------
# 11. SURVIVAL COUNT
# ------------------------------------------------------------

survival_counts = df["Survived"].value_counts()

plt.figure(figsize=(7, 5))

plt.bar(
    ["Did Not Survive", "Survived"],
    [survival_counts.get(0, 0), survival_counts.get(1, 0)]
)

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()


# ------------------------------------------------------------
# 12. SURVIVAL BY GENDER
# ------------------------------------------------------------

gender_survival = df.groupby("Sex")["Survived"].mean()

plt.figure(figsize=(7, 5))

plt.bar(
    gender_survival.index,
    gender_survival.values
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.ylim(0, 1)

plt.show()


# ------------------------------------------------------------
# 13. SURVIVAL BY PASSENGER CLASS
# ------------------------------------------------------------

class_survival = df.groupby("Pclass")["Survived"].mean()

plt.figure(figsize=(7, 5))

plt.bar(
    class_survival.index.astype(str),
    class_survival.values
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.ylim(0, 1)

plt.show()


# ------------------------------------------------------------
# 14. CORRELATION MATRIX
# ------------------------------------------------------------

# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)


# ------------------------------------------------------------
# 15. CORRELATION PLOT USING MATPLOTLIB
# ------------------------------------------------------------

plt.figure(figsize=(10, 8))

plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    interpolation="nearest"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

# Display correlation values
for i in range(len(correlation_matrix.columns)):
    for j in range(len(correlation_matrix.columns)):
        plt.text(
            j,
            i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 16. AGE VS FARE SCATTER PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Age"],
    df["Fare"],
    alpha=0.5
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.grid(alpha=0.3)

plt.show()


# ------------------------------------------------------------
# 17. BOX PLOT FOR AGE
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

plt.boxplot(df["Age"])

plt.title("Box Plot of Passenger Age")
plt.ylabel("Age")

plt.show()


# ------------------------------------------------------------
# 18. BOX PLOT FOR FARE
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

plt.boxplot(df["Fare"])

plt.title("Box Plot of Passenger Fare")
plt.ylabel("Fare")

plt.show()


# ------------------------------------------------------------
# 19. FINAL DATASET INFORMATION
# ------------------------------------------------------------

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nEDA completed successfully!")
```
